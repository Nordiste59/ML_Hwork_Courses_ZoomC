```python
def median_evolution(new, old):
    sentence = ''
    if new > old:
        sentence = 'The median has increased'
    elif new < old:
        sentence = 'The median has decreased'
    else:
        sentence = 'No change in median'
    return sentence
```

```python
print(median_evolution(new_median_horsepower, median_horsepower))
```

Question7: Sum of weights

```python
asia_cars = df[df.origin == 'Asia'].reset_index(drop=True)
print(asia_cars.head())
```

```python
asia_vehicle_weight_model_year = asia_cars[['vehicle_weight', 'model_year']]
print(asia_vehicle_weight_model_year)
```

```python
asia_vehicle_weight_model_year_7vals = asia_vehicle_weight_model_year.head(7)
print(asia_vehicle_weight_model_year_7vals)
```

```python
X = np.array(asia_vehicle_weight_model_year_7vals)
print(X)
```

```python
XTX = (X.T)@X
print(XTX)
```

```python
y = np.array([1100, 1300, 800, 900, 1000, 1100, 1200])
print(y)
```

```python
if np.linalg.det(XTX) != 0:
    XTX_inv = np.linalg.inv(XTX)
    W = (XTX_inv)@(X.T)@y
    print('W = {}'.format(W))
else:
    print('The XTX matrix is not inersible !!!')
```

```python
W_sum = W.sum()
print(W_sum)
```

```python
W_sum = np.sum(W)
print(W_sum)
```
